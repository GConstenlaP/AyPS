### Introducción
En este trabajo se analiza el desparramo espectral (Leakage) causado por la DFT bajo distintas condiciones de frecuencia original de la señal.

Para la experimentación se genero una señal senoidal de frecuencia original $ f_0= k_0 \cdot \frac{fS}{N} = k_0 \cdot \Delta f$ y se varió el parámetro $k_0$ para estudiar el desparramo espectral en cada caso.

Además se pidió potencia unitaria ($P = 1W$) por lo cual se utilizó Amplitud $A = \sqrt{2}$ para la señal generada.

Luego, para estudiar el desparramo espectral se estudio la densidad espectral en decibelios y se verificó la identidad de Parseval.

#### Marco Teorico
La DFT representa una señal en el dominio de la frecuencia, mediante un conjunto de frecuencias discretas, llamadas bins de frecuencia. Cada bin esta asociado a un índice k.

$f_s$,  la frecuencia de muestreo, demarca el ancho de banda total que evalúa la DFT y **N** determina el numero de de puntos de evaluación.
De acá se deduce la separación entre un bin y el consecutivo que  se denomina resolucion frecuencial y se expresa:

$$\Delta f = \frac{f_s}{N} = \frac{1}{(NTs)} $$

Donde $T_s$ es el periodo de muestro.

El numero de bin, o indice de frecuencia **k** identifica cada una de las frecuencias discretas donde la DFT evalúa la señal. La frecuencia asociada a un bin puede calcularse como:

$$f_k = k \cdot \Delta f$$ Para k= 0, 1, 2 ..., N-1

##### **Enventanado** o *Data windowing*

En el procesamiento digital de las señales, al tomar una señal $x[n]$ y muestrarla N veces, equivale matemáticamente a multiplicar la señal original por una ventana rectangular $w[n]$:

$$x'[n] = x[n] \cdot w[n]$$

con:

$w[n] = 1 $ si $0 \le n \le N-1$.

$w[n] = 0 $ en otro caso

Por esto decimos que de no aplicar una ventana específica, estamos **implicitamente aplicando una ventana rectangular.**

En el dominio frecuencial: el espectro de la ventana rectangular $w[n]$ es una funcion periódica que llamamos sinc discreta y tambien es conocida como *Kernel de Dirichlet*:
$$\vert{}W(\theta)\vert{} = \left\vert{} \frac{\sin\left(\theta \frac{N}{2}\right)}{\sin\left(\frac{\theta}{2}\right)} \right\vert{}$$

donde $\theta $ es la frecuencia digital angular y se define: $\theta = 2\pi \frac{f}{f_s}$ (rad/muestra). 

Por lo que el Kernel de dirichlet expresado en frecuencia en hertz es:

$$\vert{}W(f)\vert{} = \left\vert{} \frac{\sin\left(2\pi \frac{f}{f_s} \cdot \frac{N}{2}\right)}{\sin\left(2\pi \frac{f}{f_s} \cdot \frac{1}{2}\right)} \right\vert{} = \left\vert{} \frac{\sin\left(\pi N \frac{f}{f_s}\right)}{\sin\left(\pi \frac{f}{f_s}\right)} \right\vert{} = \left\vert{} \frac{\sin\left(\pi \frac{f}{\Delta f}\right)}{\sin\left(\frac{\pi}{N} \frac{f}{\Delta f}\right)} \right\vert{}$$

Y por propiedad de la transformada de Fourier: multiplicar en un dominio, es convolucionar en el otro. Y el espectro de la senoidal sabemos que es una delta de Dirac centrada en la frecuencia original de la señal y por lo tanto el resultado de la transformada de Fourier es convolucionar la delta de la senoidal con la sinc de la ventana.

$$X(f) = \mathcal{F}\{x[n]\} * W(f)$$

$$X(f) = \delta(f - f_0) * W(f) = W(f - f_0)$$

Y si queremos tomar muestras discretas, la DFT equivale a:
$$X[k] = \left. X(f) \right\vert{}_{f = k \Delta f} = W(k \Delta f - f_0)$$


de esta forma el espectro de la sinusoidal se ve afectado por la ventana y la energía que estaba concentrada en la delta se desparrama en un lóbulo principal de ancho $2\Delta f$  y otros secundarios de ancho $\Delta f$



Como se ve en la imagen, dependiendo de la $ f_0$ de la señal original, podemos tener distintos casos:

1. Si es multiplo entero de la resolución espectral $f_0 = k_0 \cdot \Delta f$: el lóbulo principal de la sinc coincide con el $k_0$, el bin de la frecuencia original de la señal, lo que causa que la DFT evalúe la frecuencia en el máximo de amplitud. Y los demás bins discretos ($k \neq k_0$) coinciden con los ceros de la sinc, por lo que no se obtienen mas muestras espectrales con amplitud.

2. Si no es multiplo entero $f_0 \neq k_0 \cdot \Delta f$): La frecuencia original $f_0$ no coincide con ningun bin de frecuencia, y esta se encuentre entre 2 bins, entonces La DFT evaluará en frecuencias distintas del máximo, y se espera obtener mas de una muestra espectral con amplitud para el lóbulo principal. Y ademas los bins consecutivos de la DFT no coincidiran con ceros de la sinc, entonces los lobulos secundarios también proyectaran muestras espectrales con amplitud no nula.


Para estudiar y comparar la energía de la señal en dominio del tiempo y en el dominio frecuencial usaré la **Identidad de Parseval** (Introducida en la TS1), la cual establece la conservación de la energía entre el dominio temporal y frecuencial:$$\sum_{n=0}^{N-1} \vert{}x[n]\vert{}^2 = \frac{1}{N} \sum_{k=0}^{N-1} \vert{}X[k]\vert{}^2$$

Y la potencia media a partir de Parseval:

$$
P_{\text{media}}
=
\frac{1}{N}
\sum_{n=0}^{N-1}
|x[n]|^2
=
\frac{1}{N^2}
\sum_{k=0}^{N-1}
|X[k]|^2
$$


### Desarrollo
La señal fue generada con **N** = 1000 y $f_s = 1000$. Entonces la resolución espectral es de $\Delta_f = 1hz$
Los parámetros k_0 evualuados fueron 
1. $k_0 =\frac{N}{4}$
2. $k_0 =\frac{N}{4} + 0,25$
3. $k_0 =\frac{N}{4} + 0,5$

Por lo que las frecuencias de las señales eran
1. $ f_0 = 250hz$
2. $ f_0 = 250,25hz$
3. $ f_0 = 250,5hz$

```python

import numpy as np
import matplotlib.pyplot as plt
import scipy
# %% definiciones

N = 1000
fs = 1000
n = np.arange(N)
t = n / fs

delta_f = fs / N  # resolución espectral en hz = 1
#1 f0 = k0∗fS/N   =  k0.Δf

# %% funciones
def mi_funcion_sen(vmax=np.sqrt(2), dc=0, ff=1, ph=0, nn=N, fs=fs):
    tt = np.arange(nn) / fs
    xx = dc + vmax * np.sin(2 * np.pi * ff * tt + ph)
    return tt, xx



def graficar_primer_ciclo(t, x, f, titulo):
    #implemento redondeo hacia arriba copn np.ceil
    muestras_por_ciclo = int(np.ceil(fs / f)) #muestras que componen un ciclo
    plt.figure()
    plt.plot(t[:muestras_por_ciclo + 1], x[:muestras_por_ciclo + 1],'o-')
    plt.title(titulo)
    plt.xlabel('Tiempo [s]')
    plt.ylabel('Amplitud [V]')
    plt.grid()
    plt.show()

# %% mi script
#Graficar las tres densidades espectrales de potencia (PDS's) y discutir cuál es el efecto de dicha desintonía en el espectro visualizado.

# %% 1 k0= n/4

k0 = N/4
k1 = N/4 + 0.25
k2 = N/4 + 0.5

f0 = k0 * (fs/N)
f1 = k1 * (fs/N)
f2 = k2 * (fs/N)

t0, x0 = mi_funcion_sen(dc=0, ff=f0, ph=0, nn=N, fs=fs)
t1, x1 = mi_funcion_sen(dc=0, ff=f1, ph=0, nn=N, fs=fs)
t2, x2 = mi_funcion_sen(dc=0, ff=f2, ph=0, nn=N, fs=fs)
print("la frecuencia de la señal f0 en hz es :", f0)
print("la frecuencia de la señal f2 en hz es :", f1)
print("la frecuencia de la señal f2 en hz es :", f2)
graficar_primer_ciclo(t0, x0, f0, 'Primer ciclo de Señal 1')

#graficar_primer_ciclo(t1, x1, f1, 'Primer ciclo de Señal 2')
#graficar_primer_ciclo(t2, x2, f2, 'Primer ciclo de Señal 3')
```

```python
# %% f02

# Calculo de Potencia:
# modulo_fft_X   = np.abs(scipy.fft.fft(X)[:N//2]) / N
# Max1 = np.max(modulo_fft_X)
# PSD_X   = (modulo_fft_X /Max1)**2
# PSD_X_db   = 10 * np.log10(PSD_X)

fft_0 =np.abs( scipy.fft.fft(x0)[:N//2])/N
Max1 = np.max(fft_0 )
PSD_X0   = (fft_0/Max1)**2
PSD_X0_db   = 10 * np.log10(PSD_X0)

fft_1 =np.abs( scipy.fft.fft(x1)[:N//2])/N
PSD_X1   = (fft_1/Max1)**2
PSD_X1_db   = 10 * np.log10(PSD_X1)

fft_2 =np.abs( scipy.fft.fft(x2)[:N//2])/N
PSD_X2   = (fft_2/Max1)**2
PSD_X2_db   = 10 * np.log10(PSD_X2)

frecuencias = np.arange(0, fs/2, fs/N)


plt.figure(figsize = (13,4) , dpi = 500)
plt.plot(frecuencias, PSD_X0_db, color= 'red', label  = f'f0 = {k0}hz', marker='o', markersize=1.5)
plt.plot(frecuencias, PSD_X1_db, color= 'blue', label  = f'f1 = {k1}hz', marker='o', markersize=1.5)
plt.plot(frecuencias, PSD_X2_db, color= 'yellow', label  = f'f2 = {k2}hz', marker='o', markersize=1.5)

plt.title('Gráfico 1. Desparramo espectral en potencia')
plt.xlabel("Frecuencias [Hz]")
plt.ylabel("Densidad de Potencia [dB]")
plt.grid()
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='best')
plt.show()

plt.figure(figsize = (13,4) , dpi = 500)
plt.plot(frecuencias, PSD_X0_db, color= 'red', label  = f'f0 = {k0}hz', ls = ':', linewidth=1, marker='o', markersize=1.5)
plt.plot(frecuencias, PSD_X1_db, color= 'blue', label  = f'f1 = {k1}hz', ls = ':', linewidth=1,        marker='o', markersize=1.5)
plt.plot(frecuencias, PSD_X2_db, color= 'yellow', label  = f'f2 = {k2}hz', ls = ':', linewidth=1,      marker='o', markersize=1.5)


plt.title('Gráfico 2. Desparramo espectral en potencia')
plt.xlabel("Frecuencias [Hz]")
plt.ylabel("Densidad de Potencia [dB]")
plt.grid()
plt.grid(True, linestyle='--', alpha=0.5)
plt.ylim(-100, 5)
plt.legend(loc='best')
plt.show()


plt.figure(figsize = (13,6) , dpi = 500)

plt.plot(frecuencias, PSD_X0_db, ls = ':', color= 'red', label  = f'f0 = {k0}hz', linewidth=1, marker='o', markersize=4)
plt.plot(frecuencias, PSD_X1_db, ls = '-.', color= 'blue', label  = f'f1 = {k1}hz', linewidth=1,        marker='o', markersize=4)
plt.plot(frecuencias, PSD_X2_db, ls = '-.', color= 'yellow', label  = f'f2 = {k2}hz', linewidth=1,      marker='o', markersize=4)


plt.title('Gráfica 3. Zoom de Desparramo espectral en potencia')
plt.xlabel("Frecuencias [Hz]")
plt.ylabel("Densidad de Potencia [dB]")
plt.grid()
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlim(240, 260)
plt.xticks(np.arange(240, 261, 1))
plt.ylim(-15, 5)
plt.legend(loc='best')
plt.show()
```

En las gráficas se puede apreciar que por un lado, para la señal roja de frecuencia 250hz casi no hay desparramo espectral y la potencia se concentra en el bin K0 = 250hz.

Para las otras señales, amarilla y azul, cuyas frecuencias solo difieran de una pequeña fracción, con frecuencias $f_0 = 250,25hz$ y $f_0 = 250,5hz$ se produce desparramo espectral considerable. Y se puede observar que este desparramo es aun mayor para la señal cuya frecuencia amarilla de $f_0 = 250.hz$ que se encuentra mas lejos del bin de frecuencia 250hz. Este resultado coincide con lo esperado ya que las frecuencisa originales $f_0$ de las señales y amarilla y azul no son multiplo entero de las frecuencia de los bins ($f_k$) que depende de la frecuencia de muestreo $f_s$

Con resolución espectral de 1hz, ambas frecuencias caen entre el bin k = 250: de 250hz y el bin k = 251: de 251 hz.

Entonces en ambas señales (Azul y amarilla), en el rango en frecuencia [249, 251]hz, que corresponde al Lobulo principal de la sinc,
La dft evalúa en frecuencias que no coinciden con el maximo de las señales y esto se puede ver en la tercer Gráfica, donde se ve que la señal roja marca un maximo de potencia mayor a la azul, y la azul mayor a la amarilla, ya que esta ultima todavia se encuentra mas lejos del bin.

Ademas en las señales Azul y amarilla, la dft esta evaluando frecuencias donde caen lobulos secundarios, lo que muestra el efecto del desparramo espectral y devuelve resultados de optencia en frecuencias distintas de la frecuencia de la señal original de entrada.


### Parte B: Verificación de potencia unitaria
En este apartado, usaré la identidad de Parseval para demostrar que a pesar de que los gráficos hayan devuelto distinta distribución de potencia con máximos distintos la potencia media de las señales es la misma:

```python
# %% parte B identidad de parseval

#potencia en el dominio del tiempo:
#potencia media

pot0t= np.mean(x0**2) 
pot1t= np.mean(x1**2) 
pot2t= np.mean(x2**2) 

#potencia en el dominio de la frecuencia:
    
pot0f= np.sum(np.abs(np.fft.fft(x0)**2)) / N**2
pot1f= np.sum(np.abs(np.fft.fft(x1)**2)) / N**2
pot2f= np.sum(np.abs(np.fft.fft(x2)**2)) / N**2


print(f"\nSeñal 1: f0 = {f0}hz")
print(f'Potencia en dominio temporal: {pot0t}')
print(f'Potencia en dominio frecuencial: {pot0f}')


print(f"\nSeñal 2: f0 = {f1}hz")
print(f'Potencia en dominio temporal: {pot1t}')
print(f'Potencia en dominio frecuencial: {pot1f}')

print(f"\nSeñal 3: f0 = {f2}hz")
print(f'Potencia en dominio temporal: {pot2t}')
print(f'Potencia en dominio frecuencial: {pot2f}')
```

Los datos arrojados por el programa indican que la potencia calculada en el dominio frecuencial y temporal es exactamente igual en cada señal.
Esto muestra que si bien hay instancias donde la DFT calcula menor amplitud, esta falta de potencia la compensa con los lobulos secundarios de forma que siempre se mantiene la igual de la identidad de Parseval y siuempre es equivalente la energia entre ambos dominios.
Así podemos afirmar que el enventanado no crea o destruye energía, solo la desparrama p distribuye en mas frecuencias y por esto se puede apreciar el efecto de desparramo espectral en las gráficas de densidad de potencia espectral.


### Parte C: Zero Padding

A continuación se repetira el Experimento utilizando la tecnica de **Zero Padding** que consiste en agrandar de tamaño un vector o matriz, exteniendo una o mas de las dimensiones y rellenando los nuevos espacios con ceros.

El zero padding tiene dos aplicaciones principales en procesamiento de señales:
1. Evitar el Aliasing circular causado al convolucionar una señal con otra en el dominio temporal.

Por Propiedad de transformada de Fourier
$$y(t) = x(t)∗h(t)\quad \xleftrightarrow{\text{FT}}\quad Y(f) = X(f)H(f)$$

Esta propiedad se mantiene para la **DTFT**, la transformada de fourier de tiempo discreto:

$$y[n] = x[n] * h[n]\quad \xleftrightarrow{\text{DTFT}} \quad Y(e^{j\omega}) = X(e^{j\omega})H(e^{j\omega})$$

Sin embargo la **DFT** transforma la representacion de la señal al dominio frecuencial, implicando periodicidad de la señal (de Período = N muestras) y entonces la propiedad no se mantiene y la propiedad que aplica en este caso es con la convolución circular de tamaño N.

$$y[n] = x[n] \circledast_N h[n] \xleftrightarrow{\text{DFT}}\quad Y(k) = X(k)H(k)$$
Por lo tanto si se calcula el producto de la DFT de dos señales para obtener Y(k), al hacer la DFT inversa, se obtiene la convolución circular que produciria **Solapamiento temporal** (*Circular Aliasing*) y para aplacar esto, es que se utiliza la técnica de Zero Padding agregando una cantidad de ceros que coincida con el tamaño del vector de la convolución lineal ($N_l$), que depende de las muestras de la señal x ($N_x$) y de la señal h ($N_h$)
$$
N_l \geq N_x + N_h - 1
$$
Y de esta forma la convolución circular coincide con la convolución lineal, permitiendo calcularla mediante la DFT

2. Al analizar el espectro de densidad de potencia frecuencial, el caso de este trabajo; cuando una o mas componentes frecuenciales ($f_0$) no coincide con la frecuencia $f_k$ de ninguno los bins y se produce desparramo espectral.

    Por eso voy a utilizar la técnica a continuación y analizar los resultados obtenidos:
   

A continuación se realiza el Zero padding con 10.000 muestras totales (M), concatenando las funciones de las señales originales con vectores de 9000 ceros.
Es importante diferenciar que con el Zero Padding, al calcular la DFT sobre un total de muestras mas alto, modificamos la densidad de muestras espectrales que obtenemos, y esto es distinto a la resolución espectral, que depende de la cantidad de muestras N tomadas sobre la señal y del periodo entre ellas $T_s$ que, al hacer Zero padding, permanece constante.

```python
# %% parte C identidad de parseval

#forma manual
M = N + 9*N #10000
vector_pad = np.zeros(N*9)

x0_pad = np.append(x0, vector_pad) #vector de 10.000 espacios
#otra forma: np.concatenate((x0, vector_pad))
# otra forma: x0_padded = np.pad(x0, (0, M - N))


x1_pad = np.append(x1, vector_pad)
x2_pad = np.append(x2, vector_pad)


muestreo_pad = fs / M  # misma formula que delta_f pero conceptualmente es algo distinto
frecuencias_pad = np.arange(0, fs/2, fs/M) # de 0 a 500 con paso de 0,1 hz


# Otra forma mas directa era:fft_x0_pad = np.abs(np.fft.fft(x0, n=M)[:M // 2]) / N
fft_0_pad =np.abs( scipy.fft.fft(x0_pad)[:M//2])/N # se divide por N porque solo tengo 1000 valores distintos de 0
Max1 = np.max(fft_0 )
PSD_X0_pad   = (fft_0_pad/Max1)**2
PSD_X0_db_pad   = 10 * np.log10(PSD_X0_pad)



fft_1_pad =np.abs( scipy.fft.fft(x1_pad)[:M//2])/N # se divide por N porque solo tengo 1000 valores distintos de 0
PSD_X1_pad   = (fft_1_pad/Max1)**2
PSD_X1_db_pad   = 10 * np.log10(PSD_X1_pad)


fft_2_pad =np.abs( scipy.fft.fft(x2_pad)[:M//2])/N # se divide por N porque solo tengo 1000 valores distintos de 0
PSD_X2_pad   = (fft_2_pad/Max1)**2
PSD_X2_db_pad   = 10 * np.log10(PSD_X2_pad)


plt.figure(figsize = (12,5 ), dpi = 500)
plt.scatter(frecuencias_pad, PSD_X0_db_pad, label = 'DFT con Zero Padding ko = 0, f0 = 250hz', marker='o', s=10)
plt.plot(frecuencias, PSD_X0_db, label = 'DFT sin Zero Padding ko = 0, f0 = 250hz', marker='o', markersize=4, color = "orange")
plt.title('Gráfico 4. Densidad Espectral con Zero Padding')
plt.grid()
plt.xlabel("Frecuencias [Hz]")
plt.ylabel("Densidad de Potencia [dB]")
plt.ylim(-80, 10)

plt.legend()
plt.show()


plt.figure(figsize = (12,5 ), dpi = 500)
plt.title('Gráfico 5. Zoom: Densidad Espectral con Zero Padding')
plt.plot(frecuencias_pad, PSD_X0_db_pad, label = 'DFT con Zero Padding ko = 0, f0 = 250hz', marker='o', markersize=4)
plt.plot(frecuencias, PSD_X0_db, label = 'DFT sin Zero Padding ko = 0, f0 = 250hz', marker='o', markersize=4)

plt.grid()
plt.xlabel("Frecuencias [Hz]")
plt.ylabel("Densidad de Potencia [dB]")
plt.xlim(240,260)
plt.xticks(np.arange(240, 261, 1))
plt.tight_layout()

plt.legend()

plt.show()
```

En esta gráfica (Zoom), a partir de la obsrevacion se puede entender el por que del resultado de la grafica de densidad espectral de la señal de 250 hz visto anteiormente.

El espaciado de la grilla se diseño para coincidir con los bins de frecuencia de la dft de resolucion de 1hz, que según lo esperado se mantuvo constante, esto puede observarse también en el ancho del lóbulo principal de la ventana, que equivale al doble de la resolucion espectral.
Como se esperaba, los bins de frecuencia (Puntos naranja) coinciden justo con los ceros de los lobulos secundarios de la sinc generada por el enventanado rectangular implicito. Una aproximación de la sinc puede verse en el grafico debido a tener un muestreo espectral mas denso, cada 0,1hz.

Este es el caso sintonizado y a continuación se graficaran los casos no sintonizados donde:

$$f_0 \neq f_k = k \cdot \Delta f$$

Se puede ver que esto ocurre porque el maximo de la sinc de la señal de 250.25hz no coincide con los binds de la DFT, en este caso el bin mas proximo al máximo es el 250 y el segundo el 251, y por eso se obtienen 2 amplitudes de potencia para el lobulo principal y uno mas por cada lobulo secundarios y así se produce el desparramo espectral.

### Frecuencia de 250.5hz


```python


plt.figure(figsize = (12,5 ), dpi = 500)
plt.scatter(frecuencias_pad, PSD_X2_db_pad, label = 'DFT con Zero Padding ko = 0, f0 = 250.5hz', marker='o', s=10)
plt.plot(frecuencias, PSD_X2_db, label = 'DFT sin Zero Padding ko = 0, f0 = 250.5hz', marker='o', markersize=4, color = "orange")
plt.title('Densidad Espectral con Zero Padding')
plt.grid()
plt.xlabel("Frecuencias [Hz]")
plt.ylabel("Densidad de Potencia [dB]")
plt.ylim(-80, 10)

plt.legend()
plt.show()


plt.figure(figsize = (12,5 ), dpi = 500)
plt.title('Zoom: Densidad Espectral con Zero Padding')
plt.plot(frecuencias_pad, PSD_X2_db_pad, label = 'DFT con Zero Padding ko = 0, f0 = 250hz', marker='o', markersize=4)
plt.plot(frecuencias, PSD_X2_db, label = 'DFT sin Zero Padding ko = 0, f0 = 250hz', marker='o', markersize=4)

plt.grid()
plt.xlabel("Frecuencias [Hz]")
plt.ylabel("Densidad de Potencia [dB]")
plt.xlim(240,260)
plt.xticks(np.arange(240, 261, 1), rotation=90)
plt.ylim(-52, 10)

plt.tight_layout()

plt.legend()

plt.show()
```

Para la señal de frecuencia 250.5 hz podemos ver que el máximo de la sinc se encuentra justo en el medio entre los bins 250 y 251, por lo que se producen 2 maximos de igual magnitud en el lobulo principal y tambien los bins de frecuencia de la Dft coinciden exactamente con los maximos de los lobulos secundarios de la sinc, nuevamente causando desparramo espectral.
Tambien se puede hacer un análisis acerca de las magnitudes obtenidas, en este caso la amplitud obtenida cae siempre en los máximos por lo tanto para esta sinc, el desparramo es máximo.



### Conclusion
A partir de las experimentaciones realizadas se pueden hacer algunas conclusiones:
Al tomar N muestras, se aplica una ventana implícita llamada ventana rectangular.
La resolución espectral determina la frecuencia de los bins y asi podemos saber, al menos en el caso de una senoidal si obtendremos desparramo espectral o amplitud concentrada en una componente frecuencial. A pesar del desparramo espectral, la potencia media o energía de una señal, se mantiene constante. El zero padding no influye en la resolucion espectral, solo en la densidad de muestreo, aumentando esta ultima, lo cual nos permite observar el resultado de aplicar una ventana a una señal.
