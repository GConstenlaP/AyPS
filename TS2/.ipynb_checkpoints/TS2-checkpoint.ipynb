En el presente trabajo se sintetiza una señal sinusoidal, se simula ruido de señal y se analiza el proceso de digitalizacion de dicha señal a traves de la simulación de un conversor analogico digital ADC de B bits.


### Marco Teórico
El proceso de conversion analógico digital, constituye ademas del muestreo temporal visto anteriormente, la cuantización de amplitud. La cuantización introduce el error de cuantización $e$, acotado por:
$$e \in [-q/2 : q/2]$$
donde $q$ es el paso de cuantización o resolución del conversor, y representa la distancia entre un nivel y el siguiente de los posibles valores en amplitud (V) de las muestras de la señal.

El paso de cuantizacion es calculado por:

$$q= \frac{2V_f}{2^B}$$

Donde $Vf$ es el limite del rango de escala completa (full scale) del ADC: $[-V_F, +V_F]$ y representa el limite máximo de tension que el ADC soporta. B representa los B-bits, es decir la cantidad de bits de resolución del ADC.

El error de cuntización puede expresarse estadísticamente como la *potencia media del ruido de cuantizacion* Prq: $$Prq = \frac {q^2}{12}$$


Luego para modelar un posible ruido analógico de una señal de entrada real, se introduce una variable aleatoria de distribucion normal que representa dicho ruido (*R*) expresado:

$$
R \sim \mathcal{N}(\mu,\sigma^2)
$$

Donde la media es $\mu = 0$ y la varianza: $ \sigma^2 = Pra$. La varianza usada $Pra$ equivale a la potencia de la señal de ruido y esta se definio tomando de referencia una escala $K$ y la potencia del error de cuantización $Prq$ según:

$$Pra = Prq * K$$

De forma que la potencia del ruido analógico queda definido por la potencia del error de cuantización y se puede analizar el impacto del ruido analógico en relacion al error de cuantizacion generado por el ADC.


## Desarrollo
Importo librerias a utilizar, defino algunos parametros e introduzco la señal original: senoidal de energia normalizada y varianza unitaria y por lo tanto ampltud: $A1 = squareroot(2)$:

```python

import numpy as np
import matplotlib.pyplot as plt
import scipy


N = 1000            #muestras
fs = 1000           #frecuencia de muestreo del ADC


f0 = fs/N           #frecuencia señal FUNCION DE FS Y N 
delta_f = fs / N    #f0 = fs/N = delta_f, densidad espectral
    
n = np.arange(N)

def mi_funcion_sen(vmax=1, dc=0, ff=1, ph=0, nn=N, fs=fs):

    tt = np.arange(nn) / fs

    xx = dc + vmax * np.sin(2 * np.pi * ff * tt + ph)

    return tt, xx


A1 = np.sqrt(2)
T, X = mi_funcion_sen(vmax=A1, dc=0, ff=f0, ph=0, nn=N, fs=fs)
```

Introduzco Ruido Incorrelado y Gaussiano, para el cual uso np.random.normal que es definido a partir de la potencia del error de cuantizacion escalado por K = 1, de esta forma genero un ruido analogico con igual potencia al ruido de cuantización.



Luego defino La señal analogica de entrada (XR) como la suma de la señal original (X) y el ruido generado (R) y además realizo el proceso de cuantizacion que es simulado obteniendo el valor mas cercano a la señal y es devuelto a al escala de voltios multiplicando el resultado por el paso de cuantizacion (q) y lo grafico.


#### Punto A: SEÑAL 1

```python
def gen_noise_normal(media=0, varianza=0.1, nn=N):

    desv_est = np.sqrt(varianza) #sigma

    ruido_normal = np.random.normal(media, desv_est, nn)
    return ruido_normal

    
Vf = 2
#paso de cuantizacion q, con B bits = 4 es q = 0,25V
B = 4
q = (2*Vf) / (2**B)         # q paso de cuantizacion: q = 2*Vf / 2**B........ q(Resolucion del ADC o Paso de cuantizacion)= Rango total/ Niveles

Prq = (q**2) / 12       
K = 1
Pra = Prq * K # #Potencia de Ruido analogico.
R = gen_noise_normal(varianza=Pra)

# Señal con Ruido
XR = X + R

XRQ = np.round(XR / q) * q


plt.figure(figsize=(10, 4), dpi=500)
plt.plot(T, XR, color = 'red', label = 'Señal con Ruido Analógico', linewidth=3)
plt.plot(T, X, color = 'black', label = 'Señal original', linewidth=2)
plt.plot(T, XRQ, color = 'cyan', label = 'Señal OUT del ADC', linewidth=0.4, marker='o', markersize=1, markeredgecolor='blue', markeredgewidth=0.5)
#alpha es la opacidad, 
plt.title('Gráfica con comparacion de señales. ADC de 4 Bits')
plt.xlabel('Tiempo [segundos]')
plt.ylabel('Amplitud [V]')
plt.grid()
plt.legend(loc='best')
plt.show
```

En este Gráfico se comparan la señal original, la Señal con Ruido Analógico y La señal obtenida a la salida del ADC de 4 bits, es decir la señal con ruido analóglico luego del proceso de cuantizacion, lo que conlleva otro error en la medición llamado ruido de cuantización. 

El efecto de la cuantizacion en la señal de entrada se puede apreciar en el gráfico, donde se ve que la amplitud de la señal deja de tomar valores continuos y ahora puede tomar valores discretizados (Puntos en azul) que corresponden a ciertos niveles de tensión )=, que son los niveles de cuantización separados por paso de cuantización (q = 0,25V). Esto causa que la señal digitalizada presente saltos entre los distintos niveles.

Luego, gracias al Ruido analógico se ve que aparecen instancias de pequeños intervalos en las cuales primero se toman ciertos valores de señal, luego se toman valores pertenecientes a un nivel mayor y luego a un nivel menor, de forma oscilatoria, esto se puede observar a partir de las lineas celestes generadas con el programa. Esto se hace evidente en el máximo y mínimo principales del ciclo de la señal, en los cuales, de ser por la señal original, la señal luego de la cuantización del ADC tomaria por cierto intervalo un continuo del valor de q = 1,5V y q = -1,5V, para el máximo y mínimo respectivamente. Pero gracias al ruido analógico se puede observar que el valor de La señal de Salida en estas regiones oscila entre 1,5V y 1,25V.

Por lo tanto en el gráfico se pueden observar 2 fuentes de error: el error de cuantizacion (Eq) y el error generado por el ruido analogico.
Cómo aclaración, en la gráfica no llegan a observarse todos los niveles porque la amplitud de la señal (1.5V) es menor que el rango de entrada analógica del ADC (±VF=2) por lo que la señal no toma todos los valores posibles que el ADC cuantiza.

### Gráfica de Densidad espectral

```python

# A partir de Ahora graficare solo el rango de (0, fs/2)
frecuencias = np.arange(0, fs/2, fs/N)

# FFTs normalizadas acotadas a Nyquist (0 a fs/2)
modulo_fft_X   = np.abs(scipy.fft.fft(X)[:N//2]) / N
modulo_fft_XR  = np.abs(scipy.fft.fft(XR)[:N//2]) / N
modulo_fft_XRQ = np.abs(scipy.fft.fft(XRQ)[:N//2]) / N

#Normalizacion de Señal respecto a amplitud maxima
#Escalo para que el pico max de amplitud de la señal original coincida con el 0
Max1 = np.max(modulo_fft_X)

# calculo de Potencia 
#La potencia de una componente espectral es la amplitud en v, al cuadrado

PSD_X   = (modulo_fft_X /Max1)**2
PSD_XR  = (modulo_fft_XR /Max1)**2
PSD_XRQ = (modulo_fft_XRQ /Max1)**2

#Conversión a dB (10 * log10 para Potencia)
PSD_X_db   = 10 * np.log10(PSD_X)
PSD_XR_db  = 10 * np.log10(PSD_XR)
PSD_XRQ_db = 10 * np.log10(PSD_XRQ)


#El piso analogico R es el valor medio del ruido (analogico) gaussiano que se agrega a la señal  antes de pasar por el adc
# El piso Digital Eq es el valor medio de potencia del error de cuantizacion Eq = XRQ - XR

#Error de cuantización
eq = XRQ - XR
modulo_fft_eq_norm = np.abs(scipy.fft.fft(eq)[:N//2]) / N
PSD_eq    = (modulo_fft_eq_norm / Max1)**2
PSD_eq_db = 10 * np.log10(PSD_eq)


piso_analog  = 10 * np.log10(np.mean(PSD_XR[2:]))
piso_digital = 10 * np.log10(np.mean(PSD_eq[2:]))


plt.figure()
plt.plot(frecuencias, PSD_X_db, color ='black', alpha = 1, zorder = 5, label = 'PSD de Señal original')
plt.plot(frecuencias, PSD_XR_db, color = 'Red', alpha = 0.7, zorder = 4, label = 'PSD de Señal con Ruido')
plt.plot(frecuencias, PSD_XRQ_db, color = 'Blue', alpha = 0.7, zorder = 6, label = 'PSD de Señal con Ruido cuantizada')

plt.plot(frecuencias, PSD_eq_db , color = 'Yellow', alpha = 0.7, zorder = 4, label = 'PSD del Error de cuantizacion')

#lineas de piso de Ruido
plt.axhline(piso_analog, color='orange', linestyle='--', label=f'piso analog. ({piso_analog:.1f} dB)', lw = 2, zorder = 8)
plt.axhline(piso_digital, color='darkcyan', linestyle=':', label=f'piso digital ({piso_digital:.1f} dB)', lw = 4,zorder = 8)

plt.title("Densidades de Potencia de Señales")
plt.ylabel("Densidad de Potencia [dB]")
plt.xlabel("Frecuencia [Hz]")
plt.grid(True, linestyle='--', alpha=0.5)
plt.ylim(-85, 10)
plt.legend(loc='best')
plt.axhline(0,color="black")
plt.axvline(0,color="black",linewidth=1)

```

A partir de esta gráfica se puede obtener la disutribución frecuencial de la señal y su contribución en potencia.

Se puede ver que la componente frecuencial en 1 hz de las señales: Señal con Ruido Analógico (Rojo), de salida cuantizada (Azul) y original (Negro) tienen un pico de potencia. Esto es porque 1hz es la frecuencia original de la señal. La señal ademas fue generada con Potencia unitaria de 1Watt, por lo tanto el pico se encuentra en 0db. (10log10(1W) = 0db)

El piso analogico es una recta con valor igual al valor medio del ruido analogico gaussiano que se agrega a la señal antes de pasar por el adc.
El piso Digital es una recta con valor igual al valor medio de potencia del error de cuantizacion Eq = XRQ - XR

El cálculo de ambos promedios de ruidos resultó muy próximo y es acord a lo esperado ya que para este experimento se generó la señal de ruido analógico con Varianza equivalente a su potencia media ($Pra$) y se utilizó K=1, por lo que se tenia misma potencia media de Ruido Analógico y Ruido digital. $$Pra = Prq * K$$ 


### Analisis del error de cuantizacion
A continuación, se someteran bajo distintos tipos de análisis los datos obtenidos, para poder determinar un resultado acerca de la naturaleza del ruido de cuantizacion. 


#### Histograma del ruido de cuantizacion
El histograma es una forma de representar la probabilidad experimental del error de cuantizacion que introduce el ADC, viendo el grafico podemos ver si se aproxima a una variable aleatoria de distribucion uniforme o si se aleja.
Para una distribucion uniforme ideal, todas las barras deberian tener la misma altura, es decir deberia haber la misma cantidad de muestras en cada valor de amplitud de error,


El modelo teorico espera que el error de cuantizacion (eq) se encuentre estrictamente en el intervalo: [-q/2, q/2]


```python
# %% Error del ruido de cuantizacion
# Histograma del error
#Error de cuantizacion eq = XRQ - XR


plt.figure()
plt.hist(eq, edgecolor='black')
plt.title('Histograma del ruido de cuantización')
plt.xlabel('Magnitud del error del Ruido de cuantizacion')
plt.ylabel('Cantidad de muestras')
plt.grid()
plt.show()

#El histograma es una forma de representar la probabilidad experimental del error de cuantizacion que introduce el ADC, viendo el grafico podemos ver si se aproxima a una variable aleatoria de distribucion uniforme o si se aleja, pero no es suficiente para determinar.
# Para una distribucion uniforme ideal, todas las barras deberian tener la misma altura, es decir deberia haber la misma cantidad de muestras en cada valor de amplitud de error,


#el modelo teorico espera: que nq este entre -q/2 y q/2

# reviso minimo y maximo de eq, no deberia encontrar valoresmuy lejos de ±qq/2 
print("El Valor mínimo del error en amplitud:", np.min(eq)) #en V
print("El Valor máximo del error en amplitud:", np.max(eq))


# Con Q = 0,25, el rango del erri es -q/2 y q/2 = [-0,125, 0,125]

#Varianza

var_real = np.var(eq)
var_teorica = (q**2) / 12
print("El paso de cuantizacion en volts es:", q) # V
#print("Varianza experimental:", var_real)
#print("Varianza teórica:", var_teorica)


```

Se puede ver en el histograma que si bien hay muestras muy cerca de los limites del modelo teórico, en este caso -0,125V y 0,125V, efectivamente el error de cuantizacion se encuentra estrictamente dentro del intervalo esperado. 
Ademas se puede apreciar cierta tendencia en el histograma, con las magnitudes de errores aproximandose a 100 muestras para cada barra. Probablemente con un numero mayor de muestras, esta tendencia se aproxime aún más y el error relativo sea menor.
A pesar de esto, el histograma nos da información sobre la distribución de amplitudes, pero no información acerca de la dependencia temporal entre muestras consecutivas, por lo que este metodo no es suficiente para determinar si el Ruido de cuantización se trata de una variable uniforme.


#### AUTOCORRELACION
La autocorrelacion es una herramiente que mide la relación que tiene una variable numérica consigo misma en diferentes momentos del tiempo o del espacio, en este caso una variable comparada consigo misma desplazada en el tiempo por un desfase (lag). Y permite evaluar si el ruido en dado instante guarda relación con sus valores futuros.

Un **Ruido Blanco** se define como una señal aleatoria con muestras incorreladas en el tiempo.


```python
# %% Autocorrelacion 
autocorr = np.correlate(eq, eq, mode='full')
lags = np.arange(-N + 1, N)

# Normalización respecto al pico central (lag 0)
autocorr_norm = autocorr / np.max(autocorr)

plt.figure(figsize=(9, 4))
plt.scatter(lags, autocorr_norm, color= 'blue', s = 15)
plt.title('Autocorrelación del Ruido de Cuantización (eq)')
plt.xlabel('Lag [muestras]')
plt.ylabel('Autocorrelación Normalizada')
plt.grid(True)
plt.ylim(-0.2, 1.1)
plt.show()

autocorr_potencia = np.correlate(eq, eq, mode='full') / N

# La potencia medida corresponde al lag central
potencia_medida = autocorr_potencia[N - 1]
# Potencia teórica del ruido de cuantización 
potencia_teorica = (q**2) / 12

# Error relativo porcentual
error_relativo = np.abs(potencia_medida - potencia_teorica) / potencia_teorica * 100

print(f"Potencia medida: {potencia_medida:.6f}")
print(f"Potencia teórica: {potencia_teorica:.6f}")
print(f"Error relativo: {error_relativo:.2f}%")
```

La funcion de autocorrelación teórica de un ruido blanco corresponde a una funcion con un unico impulso en k = 0 y valor nulo para cualquier otro caso, con amplitud igual a su varianza:  $\sigma_e^2 = \frac{q^2}{12}$ en este caso igual a la potencia media.

En la gráfica se observa que esto se cumple ya que hay un valor destacable aislado en la muestra k = 0 y los valores para otros k oscilan entorno al cero.
Ademas al evaluar la amplitud en k = 0m se obtuvo un valor con apenas aproximadamente 5% de error relativo con el valor teórico.
Estos 2 hechos confirman que el ruido de cuantización es un ruido blanco.

#### Test de Kolmogórov-Smirnov

El Test de Kolmogórov-Smirnov es un método estadístico para determinar si por un lado, la muestra de datos evaluada pertenece a una distribucion de probabilidad teorica determinada o  por otro lado si la muestra no se ajusta a dicha distribución o si las muestras corresponden a distribuciones distintas.
Calcula la función de distribución empírica y la compara con la función de distribución acumulada teórica para determinar un resultado.

En este análisis, los posibles resultados son: 

Hipotesis Nula, H0: Los datos del error de cuantizacion (EQ) siguen una distribución uniforme en el intervalo [q/2 , -q/2]

Hipótesis Alternativa , H1: Los datos no siguen una distribución uniforme en dicho intervalo

Para determinar el resultado se define el P-Valor y:

Si p > 0,05, no se rechaza H0

Si p ≤ 0,05, se rechaza H0



```python
from scipy.stats import kstest

a = -q / 2
b = q / 2
ancho_intervalo = b - a # = q

resultado_ks = kstest(eq, 'uniform', args=(a, ancho_intervalo))
print("p-value:", resultado_ks.pvalue)


alpha = 0.05
if resultado_ks.pvalue > alpha:
    print("No se rechaza H0. El error de cuantizacion se comporta como una distribución Uniforme [-q/2, q/2].")
else:
    print("Se rechaza H0. El error de cuantizacion NO sigue una distribución uniforme.")
    
```

Como se puede observar en los gráficos, para el caso de k = 0,1, el ruido generado con menor potencia, la linea roja se encuentra muy cercana a la señal original y esto causa que se formen "pisos" continuos separados por q = 0,25V en los cuales no oscila el valor de la amplitud.

Debido a estas oscilaciones mas pronunciadas, parece que para cada muestra consecutiva de la salida del ADC, se salta de un valor a otro aleatoriamente.

```python
# %%  Segunda Grafica

# FFTs normalizadas acotadas a Nyquist (0 a fs/2)
modulo_fft_XR0  = np.abs(scipy.fft.fft(XR0)[:N//2]) / N
modulo_fft_XRQ0 = np.abs(scipy.fft.fft(XRQ0)[:N//2]) / N

modulo_fft_XR10  = np.abs(scipy.fft.fft(XR10)[:N//2]) / N
modulo_fft_XRQ10 = np.abs(scipy.fft.fft(XRQ10)[:N//2]) / N



PSD_XR0  = (modulo_fft_XR0 /Max1)**2
PSD_XRQ0 = (modulo_fft_XRQ0 /Max1)**2

PSD_XR10  = (modulo_fft_XR10 /Max1)**2
PSD_XRQ10 = (modulo_fft_XRQ10 /Max1)**2



PSD_XR0_db  = 10 * np.log10(PSD_XR0)
PSD_XRQ0_db = 10 * np.log10(PSD_XRQ0)

PSD_XR10_db  = 10 * np.log10(PSD_XR10)
PSD_XRQ10_db = 10 * np.log10(PSD_XRQ10)



#El piso analogico R es el valor medio del ruido (analogico) gaussiano que se agrega a la señal  antes de pasar por el adc
#El piso Digital Eq es el valor medio de potencia del error de cuantizacion Eq = XRQ - XR

#Error de cuantización
# k=0,1
eq0 = XRQ0 - XR0
modulo_fft_eq_norm0 = np.abs(scipy.fft.fft(eq0)[:N//2]) / N
PSD_eq0    = (modulo_fft_eq_norm0 / Max1)**2
PSD_eq0_db = 10 * np.log10(PSD_eq0)

piso_analog0  = 10 * np.log10(np.mean(PSD_XR0[2:]))
piso_digital0 = 10 * np.log10(np.mean(PSD_eq0[2:]))

# k=10
eq10 = XRQ10 - XR10
modulo_fft_eq_norm10 = np.abs(scipy.fft.fft(eq10)[:N//2]) / N
PSD_eq10    = (modulo_fft_eq_norm10 / Max1)**2
PSD_eq10_db = 10 * np.log10(PSD_eq10)

piso_analog10  = 10 * np.log10(np.mean(PSD_XR10[2:]))
piso_digital10 = 10 * np.log10(np.mean(PSD_eq10[2:]))
# grafico k=0,1

plt.figure(figsize=(13, 4))
plt.plot(frecuencias, PSD_XR0_db, color = 'Red', alpha = 0.7, zorder = 4, label = 'PSD de Señal con Ruido')
plt.plot(frecuencias, PSD_XRQ0_db, color = 'Blue', alpha = 0.7, zorder = 6, label = 'PSD de Señal con Ruido cuantizada')
plt.plot(frecuencias, PSD_eq0_db , color = 'Yellow', alpha = 0.7, zorder = 4, label = 'PSD del Error de cuantizacion')

#lineas de piso de Ruido
plt.axhline(piso_analog0, color='red', linestyle='--', label=f'piso analog. ({piso_analog0:.1f} dB)', zorder = 8)
plt.axhline(piso_digital0, color='darkcyan', linestyle=':', label=f'piso digital ({piso_digital0:.1f} dB)', lw = 4,zorder = 8)

plt.title("Señal Muestreada por ADC 4 bits k = 0,1")
plt.ylabel("Densidad de Potencia [dB]")
plt.xlabel("Frecuencia [Hz]")
plt.grid(True, linestyle='--', alpha=0.5)
# plt.ylim(-85, 10)
plt.legend(loc='best')
plt.axhline(0,color="black")
plt.axvline(0,color="black",linewidth=1)

# grafica k=10

plt.figure(figsize=(13, 4))
plt.plot(frecuencias, PSD_XR10_db, color = 'Red', alpha = 0.7, zorder = 4, label = 'PSD de Señal con Ruido')
plt.plot(frecuencias, PSD_XRQ10_db, color = 'Blue', alpha = 0.7, zorder = 6, label = 'PSD de Señal con Ruido cuantizada')
plt.plot(frecuencias, PSD_eq10_db , color = 'Yellow', alpha = 0.7, zorder = 4, label = 'PSD del Error de cuantizacion')

#lineas de piso de Ruido
plt.axhline(piso_analog10, color='red', linestyle='--', label=f'piso analog. ({piso_analog10:.1f} dB)', zorder = 8)
plt.axhline(piso_digital10, color='darkcyan', linestyle=':', label=f'piso digital ({piso_digital10:.1f} dB)', lw = 4,zorder = 8)

plt.title("Señal Muestreada por ADC 4 bits k = 10")
plt.ylabel("Densidad de Potencia [dB]")
plt.xlabel("Frecuencia [Hz]")
plt.grid(True, linestyle='--', alpha=0.5)
# plt.ylim(-85, 10)
plt.legend(loc='best')
plt.axhline(0,color="black")
plt.axvline(0,color="black",linewidth=1)
```

Lo primero que se puede observar es el tipo de Ruido que domina en cada señal: En el primer gráfico (k=0,1) el Ruido de cuantización tiene aproximadamente 10 db mas de potencia que el ruido analógico. 

En la grafica de k=10 ocurre lo contrario, ahora el ruido analógico domina la señal y tiene aproximadamente 10 db mas que el ruido digital, incluso se puede observar que la linea de potencia de la señal cuantizada (Salida del ADC :azul) sigue fielmente a la linea de la señal con Ruido(Rojo), esto se debe a que ahora el error de cuantización es minimo frente a las oscilaciones dominantes del rui analógico.



### Análisis del ruido de cuantización

A continuación se someteran las señales de k=0,1 y k = 10 bajo el mismo analisis sobre su Ruido de cuantización, que la señal original de k = 1. 
#### Señal con k=0,1

```python
# %%  Histograma del error
#Error de cuantizacion eq = XRQ - XR


plt.figure()
plt.hist(eq0, edgecolor='black')
plt.title('Histograma del ruido de cuantización')
plt.xlabel('Magnitud del error del Ruido de cuantizacion')
plt.ylabel('Cantidad de muestras')
plt.grid()


# %% Autocorrelacion 
autocorr0 = np.correlate(eq0, eq0, mode='full')
lags0 = np.arange(-N + 1, N)

# Normalización respecto al pico central (lag 0)
autocorr_norm0 = autocorr0 / np.max(autocorr0)

plt.figure(figsize=(9, 4))
plt.scatter(lags0, autocorr_norm0, color= 'blue', s = 15)
plt.title('Autocorrelación del Ruido de Cuantización (eq)')
plt.xlabel('Lag [muestras]')
plt.ylabel('Autocorrelación Normalizada')
plt.grid(True)
plt.ylim(-0.2, 1.1)
plt.show()

autocorr_potencia0 = np.correlate(eq0, eq0, mode='full') / N

# La potencia medida corresponde al lag central
potencia_medida0 = autocorr_potencia0[N - 1]
# Potencia teórica del ruido de cuantización 
potencia_teorica0 = (q**2) / 12

# Error relativo porcentual
error_relativo0 = np.abs(potencia_medida0 - potencia_teorica0) / potencia_teorica0 * 100

print(f"Potencia medida: {potencia_medida0:.6f}")
print(f"Potencia teórica: {potencia_teorica0:.6f}")
print(f"Error relativo: {error_relativo0:.2f}%")

# %% Test Kolmogorov Smirnov
from scipy.stats import kstest

resultado_ks0 = kstest(eq0, 'uniform', args=(a, ancho_intervalo))
print("p-value:", resultado_ks0.pvalue)

alpha = 0.05
if resultado_ks0.pvalue > alpha:
    print("No se rechaza H0. El error se comporta como una distribución Uniforme [-q/2, q/2].")
else:
    print("Se rechaza H0. El error NO sigue una distribución uniforme.")
    
```

Histograma: En Este caso se puede observar que las muestras tomadas ya no siguen una distribucion tan uniforme. Si bien el rango de error permanece igual, ya que el paso de cuantización Q permaneció constante. Ahora Se ve que un mayor numero de muestras se concentran en ciertas magnitudes de error, no siguiendo una tendencia plana como en el histograma anterior. Y esto puede ser un indicio de que en este caso si haya correlación entre un valor y el siguiente.

El Test Kolmogorov Rechaza la hipótesis de que el ruido generado siga una distribución uniforme

Luego, el gráfico de autocorrelación si bien muestra un maximo en k = 0, ahora se presenta mayor dispersión cerca del cero, y ademas la potencia teórica respecto a la experimental segun la amplitud en k = 0 presenta un error relativo de aproximadamente 15%.

A partir de los datos observados se puede tomar en cuenta el hecho de que, ahora, al haber un ruido analógico muy reducido, el ruido de cuantización sigue una tendencia muy cercana a la señal original y no fluctua tanto por las oscilaciones del ruido analógico, por lo tanto si hay correlación entre los valores que toma.
Y por lo tanto se concluye que el Ruido de cuantización no sigue una distribución uniforme, y por lo tanto no se trata de un ruido blanco.

#### Señal con k = 10

```python
# %% k=10

#Histograma del error
#Error de cuantizacion eq = XRQ - XR


plt.figure()
plt.hist(eq10, edgecolor='black')
plt.title('Histograma del ruido de cuantización')
plt.xlabel('Magnitud del error del Ruido de cuantizacion')
plt.ylabel('Cantidad de muestras')
plt.grid()


# %% Autocorrelacion 
autocorr10 = np.correlate(eq10, eq10, mode='full')
lags10 = np.arange(-N + 1, N)

# Normalización respecto al pico central (lag 0)
autocorr_norm10 = autocorr10 / np.max(autocorr10)

plt.figure(figsize=(9, 4))
plt.scatter(lags10, autocorr_norm10, color= 'blue', s = 15)
plt.title('Autocorrelación del Ruido de Cuantización (eq)')
plt.xlabel('Lag [muestras]')
plt.ylabel('Autocorrelación Normalizada')
plt.grid(True)
plt.ylim(-0.2, 1.1)
plt.show()

autocorr_potencia10 = np.correlate(eq10, eq10, mode='full') / N

# La potencia medida corresponde al lag central
potencia_medida10 = autocorr_potencia10[N - 1]
# Potencia teórica del ruido de cuantización 
potencia_teorica10 = (q**2) / 12

# Error relativo porcentual
error_relativo10 = np.abs(potencia_medida10 - potencia_teorica10) / potencia_teorica10 * 100

print(f"Potencia medida: {potencia_medida10:.6f}")
print(f"Potencia teórica: {potencia_teorica10:.6f}")
print(f"Error relativo: {error_relativo10:.2f}%")

# %% Test Kolmogorov Smirnov
from scipy.stats import kstest

resultado_ks10 = kstest(eq10, 'uniform', args=(a, ancho_intervalo))
print("p-value:", resultado_ks10.pvalue)

alpha = 0.05
if resultado_ks10.pvalue > alpha:
    print("No se rechaza H0. El error se comporta como una distribución Uniforme [-q/2, q/2].")
```

Aca se puede ver un histograma considerablemente mas plano, lo que se alinea con una distribucion uniforme. Luego el test de Kolmogorov determino este hecho y por último la gráfica de autocorrelación representa muy proximamente a una variable incorrelada.

Los resultados para k = 10, resultan opuestos a los de k = 0,1, pero se apoyan en el análisis hecho en ese apartado: El ruido de cuantización es infulenciado por el ruido analógico. Anteriormente habiamos visto que cuando el error anlógico era pequeño, los valores tomados por la cuantización seguian mas fielmente a la señal original y esto causa que haya correlación de una muestra a la siguiente.

Ahora con un Ruido analógico potenciado, las oscilaciones generadas por este son también mas pronunciadas, esto causa que al ser cuantizada la señal, la cuantización sigue la tendencia oscilatoria aleatoria del ruido, en vez de la correlacionada de la señal original.
