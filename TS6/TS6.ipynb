### Circuito c)
#### Transferencia

Empezamos considerando al circuito como un divisor resistivo, en el cual $Z_1=R$ y $Z_2=R_L\parallel R_C$. Para este tipo de circuitos, su función transferencia es $$H_{(S)}= \frac{R_L \parallel R_C}{R+R_L \parallel R_C}$$ 

Podemos calcular $R_L \parallel R_C=\frac{SL \frac{1}{SC}}{SL+\frac{1}{SC}}$, simplificando se llega a la expresión $R_L \parallel R_C=\frac{SL}{S^2CL+1}$. Reemplazamos esto en la ecuación de divisor resistivo: $H_{(S)}=\frac{\frac{SL}{S^2CL+1}}{R+\frac{SL}{S^2CL+1}}$. Procedemos a trabajar y reorganizar esta expresión: $$H_{(S)}=\frac{SL}{S^2CLR+SL+R}=\frac{S\frac{L}{R}}{S^2CL+S\frac{L}{R}+1}$$ 

Finalmente, haremos que esta expresión se vuelva mónica: $$H_{(S)}=\frac{\frac{S}{CR}}{S^2+S\frac{1}{RC}+\frac{1}{LC}}$$

Esta es la expresión final de la función transferencia del circuito C). Comparando con (5), se puede observar que para este circuito en particular se pueden obtener las siguientes 2 condiciones: $$\frac{\omega0}{Q}=\frac{1}{RC}$$ $$\omega_0^2=\frac{1}{LC}$$
Sabiendo que R=1, $\omega_0=1$ y $Q=\frac{\sqrt2}{2}$, de estas dos condiciones se puede despejar que $C=\frac{\sqrt2}{2}$ y que $L=\sqrt2$

Para iniciar su análisis, comenzaremos por ver como se comporta el circuito a altas y bajas frecuencias. Empecemos por bajas frecuencias. En este caso, el inductor comienza a comportarse como un cortocircuito mientras que el capacitor se vuelve un circuito abierto, haciendo que la caída de tensión en salidase vuelva 0, ya que el inductor se vuelve un cable por el que pasa toda la corriente y cuya caída de tensión es 0. Para altas frecuencias, este comportamiento se invierte, siendo el capacitor el cortocircuito y el inductor el circuito abierto, provocando el mismo resultado que para altas frecuencias. Debido a esto, se puede afirmar que este circuito es un $\textbf{pasabanda}$. Haremos uso de la expresión final  para calcular manualmente como esperamos que se vea el módulo y fase de este circuito. Primero, volvemos a pasar la transferencia en función de S al dominio de j\omega:$$H_{(j\omega)}=\frac{\frac{j\omega}{SR}}{-\omega^2+\frac{j\omega}{CR} \frac{1}{LC}}$$
Calculando el módulo de esta expresión, llegamos a $$|H_{(j\omega)}|=\left|\frac{j\frac{\omega}{RC}}{-\omega^2+j\frac{\omega}{RC}+\frac{1}{LC}}\right|$$ $$|H_{(j\omega)}|=\frac{\frac{\omega}{RC}}{\sqrt{(\frac{1}{LC}-\omega^2)^2+(\frac{\omega}{RC})^2}}$$

Por su parte, la ecuación de la función transferencia se puede pensar, en términos de la amplitud y fase de P y Q, como:$$H_{j\omega}=\frac{|P_{(S)}|\angle_P}{|P_{(S)}|\angle_P}$$ 
Por lo que el diagrama de fase se obtiene restando las fases de la función $P_{(S)}$ y $Q_{(S)}$ $$\angle=\angle_P-\angle_Q=90°-tg^{-1}(\frac{\frac{\omega}{RC}}{\frac{1}{LC}-\omega^2})$$

La frecuencia de corte para este circuito resulta ser $$f_0=\frac{\omega}{2\pi}=\frac{1}{2\pi}=0,159 Hz$$

Resumiendo y reemplazando los valores tanto conocidos como calculados, llegamos a los siguientes resultados:$$H_{(S)}=\frac{\sqrt2S}{S^2+\sqrt2S+1}$$ $$|H_{(\omega)}|=\frac{\sqrt2\omega}{\sqrt{(1-\omega^2)^2+2\omega^2}}$$ $$\angle=90°-tg^{-1}(\frac{\sqrt2\omega}{1-\omega^2})$$

## <u>Conclusiones</u>
Por la respuesta del circuito a), se lo puede identificar como un filtro pasabajos. Se ve claramente que el módulo de la función transferencia se mantiene unitario para bajas frecuencias, y decrece para frecuencias $>>1$. El analisis de fase refleja que la transferencia se mantiene en el eje real positivo para bajas frecuencias, y tiende asintoticamente a valores reales negativos para $\omega>>1$; esto es debido a que el sistema tiene dos polos y ningun cero explícito que contrarreste la caida de fase aportada por cada polo.

El circuito b) responde a un filtro pasa altos de segundo orden. Su comportamiento en frecuencia confirma la atenuación de bajas frecuencias (debido al cero en el origen $z_1 = 0$) y la transferencia unitaria en altas frecuencias ($\lim_{\omega \to \infty} \vert{}H(j\omega)\vert{} = 0\text{ dB}$).Efecto de la Resonancia ($Q$):Con un factor de calidad $Q = \frac{\sqrt{2}}{2} \approx 0{,}707$, el sistema presenta una respuesta de tipo Butterworth (máximamente plana en la banda de paso). El leve sobrepico de $1{,}7\text{ dB}$ cerca de la frecuencia propia ($\omega_0 = 1\text{ rad/s}$) evidencia el intercambio de energía entre los elementos reactivos ($L$ y $C$) antes de estabilizarse en la ganancia unitaria.
Ancho de Banda y Respuesta de Fase: Se delimitó un ancho de banda efectivo a partir de la frecuencia de corte $\omega_c \approx 0{,}486\text{ rad/s}$. Asimismo, el análisis de fase muestra una transición continua desde un adelanto de $90^\circ$ en bajas frecuencias hasta alcanzar $0^\circ$ asintóticamente a altas frecuencias, lo cual es consistente con la distribución de polos y ceros en el plano complejo.

El circuito c) como un filtro pasabanda, en la transferencia, se ve que no hay término independiente en el numerador, y que el grado del denominador es mayor al del numerador, causando que la función transferencia tienda a 0 tanto en  0 como en infinito. En el gráfico de módulo se ve que solo un cierto rango de frecuencias tiene permitido el paso sin atenuación, esto debido a la frecuencia de corte.
